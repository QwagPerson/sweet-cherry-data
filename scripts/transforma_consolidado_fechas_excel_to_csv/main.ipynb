# Transforma consolidado de fechas Excel a CSV

Lee los archivos `Consolidado fechas 2024 V2.xlsx` y `Consolidado fechas 2025 V2.xlsx` desde `data/bronze/2026`, limpia nombres/fechas/metricas, agrega codigos desde los maestros y escribe los CSV limpios en `data/silver/2026`.

```python
from __future__ import annotations

from pathlib import Path
import re
import unicodedata

import pandas as pd


INPUTS = {
    2024: {
        "file": "Consolidado fechas 2024 V2.xlsx",
        "sheet": "Consolidado",
        "rename": {
            "Lugar": "lugar",
            "Variedad": "variedad",
            "UE": "ue",
            "Tratamiento": "tratamiento_entrada",
            "caida de hojas": "fecha_caida_hoja_50",
            "50 pct camara": "fecha_brotacion_camara_50",
            "50% Campo Control Salida": "fecha_brotacion_campo_50",
            "hf0 7": "hf0_mayor_7",
            "hf1 6": "hf1_mayor_6",
            "hf2 6 5": "hf2_mayor_6_5",
            "hf3 8": "hf3_mayor_8",
            "Utah 0": "utah_0",
            "Utah 1": "utah_1",
            "Utah 2": "utah_2",
            "Utah 3": "utah_3",
            "Florida 0": "florida_0",
            "Florida 1": "florida_1",
            "Florida 2": "florida_2",
            "Florida 3": "florida_3",
            "CN 0": "cn_0",
            "CN 1": "cn_1",
            "CN 2": "cn_2",
            "CN 3": "cn_3",
            "dinamico 0": "dinamico_0",
            "HGC Campo": "hgc_campo",
            "hgc camara": "hgc_camara",
        },
    },
    2025: {
        "file": "Consolidado fechas 2025 V2.xlsx",
        "sheet": "Requisitos MOD",
        "rename": {
            "Lugar": "lugar",
            "Variedad": "variedad",
            "UE": "ue",
            "Tratamiento combinado": "tratamiento_combinado",
            "Tratamiento entrada": "tratamiento_entrada",
            "50 pct caida de hoja": "fecha_caida_hoja_50",
            "50 pct camara": "fecha_brotacion_camara_50",
            "50% Campo": "fecha_brotacion_campo_50",
            "hf0 7": "hf0_mayor_7",
            "hf1 6": "hf1_mayor_6",
            "hf2 6 5": "hf2_mayor_6_5",
            "hf3 8": "hf3_mayor_8",
            "Utah 0": "utah_0",
            "Utah 1": "utah_1",
            "Utah 2": "utah_2",
            "Utah 3": "utah_3",
            "Florida 0": "florida_0",
            "Florida 1": "florida_1",
            "Florida 2": "florida_2",
            "Florida 3": "florida_3",
            "CN 0": "cn_0",
            "CN 1": "cn_1",
            "CN 2": "cn_2",
            "CN 3": "cn_3",
            "dinamico 0": "dinamico_0",
            "HGC Campo": "hgc_campo",
            "hgc camara": "hgc_camara",
        },
    },
}

DATE_COLUMNS = ["fecha_caida_hoja_50", "fecha_brotacion_camara_50", "fecha_brotacion_campo_50"]
NUMERIC_COLUMNS = [
    "ue",
    "hf0_mayor_7",
    "hf1_mayor_6",
    "hf2_mayor_6_5",
    "hf3_mayor_8",
    "utah_0",
    "utah_1",
    "utah_2",
    "utah_3",
    "florida_0",
    "florida_1",
    "florida_2",
    "florida_3",
    "cn_0",
    "cn_1",
    "cn_2",
    "cn_3",
    "dinamico_0",
    "hgc_campo",
    "hgc_camara",
]


def find_project_root() -> Path:
    for candidate in [Path.cwd(), *Path.cwd().parents]:
        if (candidate / "data" / "bronze" / "2026").exists() and (candidate / "maestros").exists():
            return candidate
    raise FileNotFoundError("No se encontro la raiz del proyecto sweet-cherry-data.")


PROJECT_ROOT = find_project_root()
BRONZE_2026 = PROJECT_ROOT / "data" / "bronze" / "2026"
SILVER_2026 = PROJECT_ROOT / "data" / "silver" / "2026"
MAESTROS = PROJECT_ROOT / "maestros"
SILVER_2026.mkdir(parents=True, exist_ok=True)


def normalize_text(value) -> str:
    if pd.isna(value):
        return ""
    text = str(value).strip().lower()
    text = unicodedata.normalize("NFKD", text).encode("ascii", "ignore").decode("ascii")
    text = text.replace("%", " pct ")
    text = re.sub(r"[^a-z0-9]+", " ", text)
    return re.sub(r"\\s+", " ", text).strip()


def clean_string(value):
    if pd.isna(value):
        return pd.NA
    text = re.sub(r"\\s+", " ", str(value).strip())
    return text if text else pd.NA


def build_lugar_lookup(lugares: pd.DataFrame) -> dict[str, str]:
    lookup: dict[str, str] = {}
    for _, row in lugares.iterrows():
        code = row["codigo"]
        names = [row["nombre"]]
        if "terminos_relacionados" in row and pd.notna(row["terminos_relacionados"]):
            names.extend(str(row["terminos_relacionados"]).split("|"))
        for name in names:
            key = normalize_text(name)
            if key:
                lookup[key] = code
    return lookup


def resolve_lugar(value, lookup: dict[str, str]):
    key = normalize_text(value)
    if not key:
        return pd.NA
    candidates = [key, re.sub(r"\\s+\\d+$", "", key).strip()]
    for candidate in candidates:
        if candidate in lookup:
            return lookup[candidate]
    for alias, code in lookup.items():
        if alias and (alias in key or key in alias):
            return code
    return pd.NA


def build_simple_lookup(df: pd.DataFrame, name_col: str = "nombre", code_col: str = "codigo") -> dict[str, str]:
    return {normalize_text(row[name_col]): row[code_col] for _, row in df.iterrows() if normalize_text(row[name_col])}


TRATAMIENTO_ENTRADA_ALIASES = {
    "control": "ECON",
    "ctrl": "ECON",
    "aba": "EABA",
    "340 ppm aba": "EABA",
    "2x200 ppm aba": "EABA",
    "2 200 ppm aba": "EABA",
    "2 x 100 ppm eth": "EETH",
    "2 5 pct zn y 150ppm eth": "EETH+EZIN",
    "2 5 pct zn y 150 ppm eth": "EETH+EZIN",
    "zn eth": "EETH+EZIN",
    "2 5l ha ciclanilida eth": "ECIC+EETH",
    "fsh": "ECIC+EETH",
    "2 5 pct zn 2 pct urea y 150 ppm eth": "EETH+EURE+EZIN",
    "zn u eth": "EETH+EURE+EZIN",
}

TRATAMIENTO_SALIDA_RULES = [
    ("1 pct ch 2 pct ac min", "SCAH+SAMI"),
    ("1 5 pct ch break", "SCAH+SCOD"),
    ("1 25 pct ch break", "SCAH+SCOD"),
    ("5 pct bioestimulante 6 pct ca no3 2", "SABI+SNDC"),
    ("ch acmin", "SCAH+SAMI"),
    ("ch break", "SCAH+SCOD"),
    ("control", "SCON"),
    ("ctrl", "SCON"),
]


def resolve_tratamiento_entrada(value):
    key = normalize_text(value)
    if not key:
        return pd.NA
    if key in TRATAMIENTO_ENTRADA_ALIASES:
        return TRATAMIENTO_ENTRADA_ALIASES[key]
    for alias, code in TRATAMIENTO_ENTRADA_ALIASES.items():
        if alias in key or key in alias:
            return code
    return pd.NA


def resolve_tratamiento_salida(value, default_control: bool = False):
    key = normalize_text(value)
    if not key:
        return "SCON" if default_control else pd.NA
    for alias, code in TRATAMIENTO_SALIDA_RULES:
        if alias in key:
            return code
    return "SCON" if default_control else pd.NA


def combine_codes(*codes):
    clean_codes = [str(code) for code in codes if pd.notna(code) and str(code).strip()]
    return "+".join(clean_codes) if clean_codes else pd.NA


def load_lookups():
    lugares = pd.read_csv(MAESTROS / "lugar.csv")
    variedades = pd.read_csv(MAESTROS / "variedad.csv")
    tratamientos = pd.read_csv(MAESTROS / "tratamiento.csv")

    lugar_lookup = build_lugar_lookup(lugares)
    variedad_lookup = build_simple_lookup(variedades)

    for _, row in tratamientos[tratamientos["tipo"].eq("entrada")].iterrows():
        TRATAMIENTO_ENTRADA_ALIASES.setdefault(normalize_text(row["nombre"]), row["codigo"])
        TRATAMIENTO_ENTRADA_ALIASES.setdefault(normalize_text(row["codigo_2"]), row["codigo"])

    return lugar_lookup, variedad_lookup


def clean_consolidado(year: int, config: dict, lugar_lookup: dict[str, str], variedad_lookup: dict[str, str]) -> pd.DataFrame:
    input_path = BRONZE_2026 / config["file"]
    df = pd.read_excel(input_path, sheet_name=config["sheet"])
    rename_lookup = {normalize_text(source): target for source, target in config["rename"].items()}
    df = df.rename(columns=lambda column: rename_lookup.get(normalize_text(column), column))
    df = df[[column for column in config["rename"].values() if column in df.columns]].copy()
    df = df.dropna(how="all")
    df.insert(0, "temporada", year)

    for column in ["lugar", "variedad", "tratamiento_entrada", "tratamiento_combinado"]:
        if column in df.columns:
            df[column] = df[column].map(clean_string)

    for column in DATE_COLUMNS:
        if column in df.columns:
            df[column] = pd.to_datetime(df[column], errors="coerce")

    for column in NUMERIC_COLUMNS:
        if column in df.columns:
            df[column] = pd.to_numeric(df[column], errors="coerce")

    df["codigo_lugar"] = df["lugar"].map(lambda value: resolve_lugar(value, lugar_lookup))
    df["codigo_variedad"] = df["variedad"].map(lambda value: variedad_lookup.get(normalize_text(value), pd.NA))
    df["codigo_tratamiento_entrada"] = df["tratamiento_entrada"].map(resolve_tratamiento_entrada)

    if "tratamiento_combinado" in df.columns:
        df["codigo_tratamiento_salida"] = df.apply(
            lambda row: resolve_tratamiento_salida(
                row["tratamiento_combinado"],
                default_control=normalize_text(row["tratamiento_combinado"]) == normalize_text(row["tratamiento_entrada"]),
            ),
            axis=1,
        )
        df["codigo_tratamiento_combinado"] = df.apply(
            lambda row: combine_codes(row["codigo_tratamiento_entrada"], row["codigo_tratamiento_salida"]),
            axis=1,
        )
    else:
        df["codigo_tratamiento_salida"] = "SCON"
        df["codigo_tratamiento_combinado"] = df.apply(
            lambda row: combine_codes(row["codigo_tratamiento_entrada"], row["codigo_tratamiento_salida"]),
            axis=1,
        )

    ordered = [
        "temporada",
        "codigo_lugar",
        "lugar",
        "codigo_variedad",
        "variedad",
        "ue",
        "codigo_tratamiento_entrada",
        "tratamiento_entrada",
        "codigo_tratamiento_salida",
        "codigo_tratamiento_combinado",
        "tratamiento_combinado",
        *DATE_COLUMNS,
        *[column for column in NUMERIC_COLUMNS if column != "ue"],
    ]
    return df[[column for column in ordered if column in df.columns]].sort_values(
        ["codigo_lugar", "codigo_variedad", "ue"], na_position="last"
    ).reset_index(drop=True)


lugar_lookup, variedad_lookup = load_lookups()
outputs = []
for year, config in INPUTS.items():
    clean_df = clean_consolidado(year, config, lugar_lookup, variedad_lookup)
    output_path = SILVER_2026 / f"consolidado_fechas_{year}_limpio.csv"
    clean_df.to_csv(output_path, index=False, encoding="utf-8")
    outputs.append((year, output_path, clean_df))

for year, output_path, clean_df in outputs:
    unresolved = {
        column: int(clean_df[column].isna().sum())
        for column in [
            "codigo_lugar",
            "codigo_variedad",
            "codigo_tratamiento_entrada",
            "codigo_tratamiento_salida",
            "codigo_tratamiento_combinado",
        ]
        if column in clean_df.columns
    }
    print(f"{year}: {len(clean_df)} filas -> {output_path.relative_to(PROJECT_ROOT)}")
    print(f"Codigos sin resolver: {unresolved}")
```
